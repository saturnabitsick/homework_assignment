**Data**

Assume the raw text file contains lines formatted like this:

`Timestamp | StationID | Temp(F) | Humidity(%) | Sensor_Status`

*Example: `2026-09-29 08:15:00 | STAT_04 | 82.5 | 45.2 | ACTIVE`*

*However, the file contains severe corruptions:*
- Some lines are completely blank.
- Some temperatures are logged as `ERROR` or `N/A`.
- The delimiter (|) might be missing.

**`clean_logs.py`**

1. Write a custom function `celsius_converter(f_temp)` that takes a Fahrenheit string/float, converts it to Celsius, and rounds it to two decimal places using `round()`.
2. Write a custom function `parse_and_clean(line)` that takes a single string line. It must split the string by the `|` delimiter, strip away excess whitespace, and check for missing data. If the line is structurally corrupted or the `Sensor_Status` is anything other than `ACTIVE`, the function should return `None`.

**Run the functions**

1. Open the input file for reading and an output file for writing.
2. Use a `for` loop to iterate through the input file line-by-line (do not load the whole file into a list, to simulate handling massive files).
3. Apply the `parse_and_clean` function.
4. For valid lines, convert the temperature using `celsius_converter`, and write the cleaned row into the output file as a standard .csv.
5. Keep a count of how many corrupted lines were skipped and print this summary when the loop finishes.

**Submission**

Create a `.gitignore` file to ignore the input and output files. Add, commit and push all changes to your Github account, make sure it's set to public (or otherwise available via link). Submit the link to your repo.

```python
import sys
sys.path.insert(0, r"C:\Projects\git\homework_assignment")
import csv

from clean_logs import parse_and_clean
from clean_logs import celsius_converter

with open(r"C:\Projects\git\homework_assignment\input\environmental_raw.txt", 'r') as file, \
     open(r"C:\Projects\git\homework_assignment\output\environmental_cleaned.csv", "w", newline="") as output_file:

    next(file)
    rejected_csv = []
    rejected_lines = []

    writer = csv.writer(output_file)
    writer.writerow(["Timestamp", "StationID", "Temp(C)", "Humidity(%)", "Sensor_Status"])

    for line in file:
        cleaned_line = parse_and_clean(line)

        if cleaned_line is None:
            print("Rejected:", line.strip())
            rejected_lines.append(line.strip())
        else:
            f_temp = cleaned_line[2]
            c_temp = celsius_converter(f_temp)
            cleaned_line[2] = c_temp
            print("Accepted:", cleaned_line)
            writer.writerow(cleaned_line)

print(f"\nCorrupted lines skipped: {len(rejected_lines)}")
```
